```python
import pandas as pd
import json
```

```python
with open("/content/export (1).json", "r") as file:
    experiments = json.load(file)

all_subject = []
for run in experiments['experiment_runs']:
  if run['status'] == 'Finalizado' and len(run['records']) > 0:
    subject_id = run['subject']
    print(subject_id)
    print("run['status']", run['status'])
    df = pd.DataFrame(eval(run['records'][-1]['data'].
                           replace('true', "True").
                           replace('false', "False").
                           replace('null', 'None')))

    df['subject_id'] = subject_id

    if df.shape[0] < 38: # Filtering data from anothe experiment
      print(df.shape)
      all_subject.append(df)

df_all_subjects = pd.concat(all_subject)
```

```python
with open("data/export.json", "r") as file:
    experiments = json.load(file)

all_subject = []
all_subject_dist = []
all_subject_blink = []
all_subject_giro = []
for run in experiments['experiment_runs']:
  if run['status'] == 'Finalizado' and len(run['records']) > 0:
    subject_id = run['subject']
    ##print(subject_id)
    ##print("run['status']", run['status'])

    data = eval(run['records'][-1]['data'].
                           replace('true', "True").
                           replace('false', "False").
                           replace('null', 'None'))
    dataLen = len(data)
    ##0-5 distance
    ##6-19 blink
    ##20-37 rotation
    if dataLen <38:
       df_dist = pd.DataFrame(data[0:6])
       df_dist['subject_id'] = subject_id
       all_subject_dist.append(df_dist)
       if dataLen > 6:
          df_blink = pd.DataFrame(data[6:19])
          df_blink['subject_id'] = subject_id
          all_subject_blink.append(df_blink)
          if dataLen > 21:
             df_giro = pd.DataFrame(data[19:])
             df_giro['subject_id'] = subject_id
             all_subject_giro.append(df_giro)

    
    # if len(data) < 38:
    #   print(data[0])
    #   print(data[1])
    #   print(len(data))
    #   print(data[34])
    #   if len(data) > 35:
    #     print(data[35])
    #     print(data[36])


df_all_subjects_dist = pd.concat(all_subject_dist)
df_all_subjects_blink = pd.concat(all_subject_blink)
df_all_subjects_giro = pd.concat(all_subject_giro)
```
